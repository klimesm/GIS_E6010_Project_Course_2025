```python
import rasterio
import numpy as np
```

```python
with rasterio.open("P4434F1_1_dem_min_hpmf.tif") as hpmf_tile:
    hpmf_array = hpmf_tile.read(1)

with rasterio.open("P4434F1_1_dem_min_label.tif") as label_tile:
    label_array = label_tile.read(1)


hpmf_patch_list = []
label_patch_list = []

raster_size = 2000
patch_size = 500

for i in range(0, raster_size, patch_size):
    for j in range(0, raster_size, patch_size):
        hpmf_patch = hpmf_array[i:i + patch_size, j:j + patch_size]
        hpmf_patch_list.append(hpmf_patch)

        label_patch = label_array[i:i + patch_size, j:j + patch_size]
        label_patch_list.append(label_patch)
```
