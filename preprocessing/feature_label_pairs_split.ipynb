```python
import rasterio
import numpy as np
```

```python
# Open the raster file containing HPMF data
with rasterio.open("P4434F1_1_dem_min_hpmf.tif") as hpmf_raster:
    hpmf_array = hpmf_raster.read(1)     # Extract band 1 as a NumPy array, where each element is a pixel value

# Open the raster file containing label data
with rasterio.open("P4434F1_1_dem_min_label.tif") as label_raster:
    label_array = label_raster.read(1)    # Extract band 1 as a NumPy array, where each element is a pixel value

# Initialize lists to store the image chips (sub-arrays) for both datasets
hpmf_chip_list = []
label_chip_list = []

# Define original raster size and chip size
original_raster_size = 2000
chip_size = 500

# Loop over the raster in steps of chip_size to create non-overlapping chips
for i in range(0, original_raster_size, chip_size):
    for j in range(0, original_raster_size, chip_size):
        
        # Extract a chip from the HPMF array
        hpmf_chip = hpmf_array[i:i + chip_size, j:j + chip_size]
        hpmf_chip_list.append(hpmf_chip)

        # Extract a chip from the label array
        label_chip = label_array[i:i + chip_size, j:j + chip_size]
        label_chip_list.append(label_chip)
```
